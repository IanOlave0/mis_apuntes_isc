# Introducción a las Redes Neuronales

Apertura del curso 2: qué son las redes neuronales, de dónde vienen, por qué dominan la actualidad y el vocabulario base con intuición visual (neurona, layers, input, hidden, output, activaciones). Todo con los mismos lentes del curso 1.

### ¿Qué es una red neuronal?

![Red neuronal profunda](../Resources/neuralNetwork.png)

Una red neuronal es un modelo formado por muchas unidades simples conectadas. La conexión con el curso 1 es directa: **una neurona con activación sigmoide es exactamente la regresión logística del Lab 06** (puntaje $z = w x + b$ aplastado a probabilidad). La novedad es apilarlas: la salida de unas alimenta la entrada de otras.

Vocabulario del diagrama:

- **Input layer** (azul): recibe los datos tal cual, una neurona por feature. No calcula nada, solo reparte.
- **Hidden layers** (verdes): capas intermedias donde se combinan los datos en patrones cada vez más abstractos.
- **Output layer** (clara): entrega la respuesta final del modelo.
- Cada flecha lleva un peso $w$: qué tan fuerte escucha una neurona a cada entrada.

### Breve historia (y por qué suenan ahora)

![Historia de las redes neuronales](../Resources/neuralNetworks_timeline.png)

- **1943**: McCulloch y Pitts proponen la primera neurona artificial (lógica con interruptores).
- **1958**: Rosenblatt crea el perceptrón, que ya aprendía pesos con ejemplos... pero solo separaba con rectas.
- **1970s**: invierno de la IA — al ver los límites del perceptrón se corta el financiamiento.
- **1986**: Hinton populariza backpropagation: por fin se pueden entrenar redes con capas ocultas.
- **2012**: una CNN gana ImageNet y nace el deep learning moderno.
- **2020s**: los transformers llevan la misma idea a texto e imágenes generativas.

¿Por qué ahora y no en los 90, si las ideas son viejas? Tres pilares se alinearon: **datos masivos** para entrenar, **cómputo barato** (GPUs) para moverlos y **algoritmos** (como backprop) que escalan. Sin esos tres, la misma red de hoy no aprendía nada.

### Lo que el curso 1 no podía (y por qué las redes sí)

Los modelos del curso 1 tienen tres techos, y las redes neuronales los rompen los tres:

1. **Rectas y planos obligados.** La regresión lineal solo dibuja rectas y planos; lo curvo exigía inventar features polinomiales a mano (lab 05).
2. **Fronteras rectas salvo truco manual.** La regresión logística separa con rectas, y solo curva si alguien crea los $x^2$ a mano, como en el anillo del microchip (lab 06).
3. **Se estancan con más datos.** Lineal y logística mejoran hasta cierto punto y luego se plantan: darles el doble de datos casi no las mueve. Las redes, en cambio, siguen mejorando mientras haya más datos, y por eso explotaron justo cuando hubo datos masivos y cómputo para moverlos.

¿Por qué las redes superan todo eso? Porque no piden features diseñadas: **las hidden layers aprenden sus propias features** (nadie les dice qué buscar), traen no-linealidad por construcción y escalan con los datos en vez de estancarse.

### La neurona (una sola)

![Neurona: predicción de demanda](../Resources/neuronExample.png)

Es una neurona de predicción de demanda. Recibe el precio $x$, calcula $a = g(w x + b)$ y devuelve la probabilidad de ser top seller.

Repaso de sus partes con nombres nuevos:

- **Entrada $x$**: el dato que recibe (aquí, el precio).
- **Pesos $w$ y sesgo $b$**: cuánto pesa cada entrada y el punto de partida.
- **Activación $a$**: la salida de la neurona tras aplicar $g$. Es un número entre $0$ y $1$.

### Capas con un ejemplo concreto: input, hidden y output

![Red de predicción de demanda](../Resources/neuralNetwork_example.png)

Ahora la red completa del mismo ejemplo. Se lee de izquierda a derecha:

- **Input layer**: el vector $x$ con 4 números reales (price, shipping cost, marketing, material). Son datos, no se calculan.
- **Hidden layer**: 3 neuronas que la red entrena para detectar ideas intermedias (aquí etiquetadas como affordability, awareness y perceived quality). Se llama *hidden* porque nadie le dice a la red qué buscar ahí: ella decide sola qué patrones le sirven.
- **Output layer**: una neurona final que combina esas 3 ideas y devuelve 1 número: la probabilidad de ser top seller.

Las etiquetas affordability/awareness/quality son una interpretación humana: lo único que la red ve son números.

### Activaciones y activation values: qué viaja por las flechas

Cada flecha lleva un número y cada círculo guarda el suyo. Esos números son las **activaciones**: la salida calculada de cada neurona para el ejemplo actual.

- La capa de entrada pasa sus 4 números tal cual (4 activation values).
- Cada neurona oculta combina esos 4 en un solo número: su activación. El vector $a$ de 3 números es la salida completa de la hidden layer.
- La neurona final combina esos 3 en 1 número: la respuesta (1 activation value).

En corto: **4 números entran, 3 intermedios se calculan, 1 número sale**. Eso es un forward pass: empujar un ejemplo de izquierda a derecha. 

### Una neurona calculada a mano

Para que la activación deje de ser abstracta, se calcula una completa a mano. Una neurona oculta recibe dos números: $x_1 = 0.5$ y $x_2 = 0.8$, con pesos $w_1 = 0.9$, $w_2 = -0.4$ y sesgo $b = 0.1$.

**Paso 1: el puntaje.** Cada entrada vota con su peso y se suma el sesgo: $z = 0.5 \cdot 0.9 + 0.8 \cdot (-0.4) + 0.1 = 0.45 - 0.32 + 0.1 = 0.23$. Nótese que el peso negativo resta: esa entrada vota *en contra* de encender la neurona.

**Paso 2: la activación.** Se aplasta el puntaje con la sigmoide: $a = g(0.23) \approx 0.557$. Un poco más de la mitad: la neurona se enciende a medias.

Eso es todo lo que hace cualquier neurona de la red, incluidas las ocultas: suma ponderada, sesgo y aplastado a probabilidad. La única diferencia en una red grande es que este cálculo se repite en cada neurona con sus propios pesos.

### Cierre y puente

Vocabulario que ya dominas: **neurona** (una unidad $a = g(w x + b)$), **layer** (grupo de neuronas del mismo nivel), **input layer** (datos), **hidden layer** (patrones intermedios que la red elige sola), **output layer** (respuesta), **activación** (el número que sale de cada neurona) y **activation values** (esos números para un ejemplo concreto).
