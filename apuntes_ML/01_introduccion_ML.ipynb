### ¿Qué es Machine Learning?
Es una rama de la IA donde las computadoras aprenden patrones a partir de datos para hacer predicciones, 
sin ser programadas con reglas explícitas.
- Programación tradicional: Datos + Reglas -> Respuestas
- Machine Learning: Datos + Respuestas -> Reglas aprendidas

##### Aplicaciones Reales

- Finanzas: Detección de fraudes y predicción de divisas.

- Salud: Diagnóstico médico por imágenes.

- Entretenimiento: Sistemas de recomendación (Spotify, Netflix). - 

- Transporte: Optimización de rutas y autos autónomos.

##### Aprendizaje Supervisado
El modelo aprende a partir de un conjunto de datos etiquetados (labeled data). Es decir, para cada ejemplo de entrada ($X$), el conjunto de datos incluye la respuesta correcta o etiqueta esperada ($y$). El objetivo es que el modelo aprenda la relación matemática entre las entradas y las salidas.
- Regresión (ej. Regresión Lineal): Predice un número continuo.
        Uso: Predecir el precio de una casa o el valor del dólar.
- Clasificación (ej. Algoritmos de Clasificación): Predice una categoría o clase.
        Uso: Determinar si un correo es Spam / No Spam o identificar imágenes.

##### Aprendizaje No Supervisado
El modelo recibe un conjunto de datos sin etiquetar (solo las variables de entrada $X$, sin ninguna respuesta $y$). El algoritmo explora los datos de forma autónoma para descubrir patrones ocultos, estructuras o agrupaciones naturales que no son evidentes a simple vista.
- Clustering / Agrupamiento (ej. K-Means): Agrupa datos según su similitud.
    Uso: Segmentación de clientes por hábitos de compra o agrupamiento de noticias.

##### Terminología común

En ML habrán terminologías y notaciones que se usan muy comunmente, algunas de ellas son:

- Training Set: La data usada para entrenar un modelo.
- x: esta será las notación que se la dará a las variables de input o feature.
- y: variable de salida o target.
- m: número de ejemplos usados para entrenar el modelo.
- (x,y): un ejemplo específico para entrenar el modelo.

- Notación $(x^{(i)}, y^{(i)})$ = (i): 

$i$: Es el número de la fila o del ejemplo (va desde $i = 1$ hasta $i = m$, donde $m$ es el total de ejemplos).

$x^{(i)}$: Es la característica de entrada (feature) del ejemplo número $i$. En el caso de casas, podrían ser los metros cuadrados de la casa número $i$.

$y^{(i)}$: Es la etiqueta o respuesta correcta (target) del ejemplo número $i$. Siguiendo el caso de casas, sería el precio real de la casa número $i$.

Para entender la notación imaginemos que tenemos una tabla donde organizamos nuesto training set de 3 ejemplos y tenemos la columna x y la columna y. En la segunda fila de nuestra tabla tenemos que $x$ = 1000 y $y$ = 25000. Entonces $(x^{(2)}, y^{(2)})$ = (2) es igual a $(1000, 25000)$ = 2. 
Donde i no es un exponencial sino que indica la fila de la tabla a la que pertenecen el conjunto de valores.

- $\mathbf{x} \longrightarrow \text{Función / Modelo } f(x) \longrightarrow \hat{y}$: donde x es el dato de entrada, $f$ o $h$ (de hipotesis) es el algoritmo o función matemática aprendida por el modelo y $\hat{y}$ es solo una predicción o estimación del resultado real.

- El símbolo del gorro ($\,\hat{}\,$) en estadística e Inteligencia Artificial denota siempre una estimación, para diferenciarla del valor real.
