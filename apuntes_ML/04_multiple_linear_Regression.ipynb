## Regresión lineal múltiple

En la vida real, una sola variable rara vez es suficiente para predecir un resultado con precisión. Por ejemplo, al predecir el precio de una casa no solo observamos su tamaño ($x$), sino también el número de habitaciones, los años de antigüedad y el número de pisos.

Cuando el modelo utiliza dos o más características de entrada (features), pasamos de la Regresión Lineal Simple a la Regresión Lineal Múltiple.

#### Nueva notación matemática

Para manejar múltiples variables sin confundirnos, adoptamos la siguiente convención matemática estándar:
- $n$: Número total de características (features). Por ejemplo, $n = 4$ si usamos (Tamaño, Habitaciones, Pisos, Antigüedad).
- $m$: Número total de ejemplos de entrenamiento (filas del dataset).
- $\mathbf{x}^{(i)}$: Un vector fila que contiene todas las características del ejemplo $i$-ésimo. Por ejemplo, $\mathbf{x}^{(2)} = [120, 3, 2, 10]$.
- $x_j^{(i)}$: El valor de la característica $j$ en el ejemplo $i$. Por ejemplo, $x_3^{(2)} = 2$ (el número de pisos de la casa 2).
- $\mathbf{w}$: Vector de pesos (weights), donde cada $w_j$ es un número que acompaña a la característica $x_j$. Se representa como $\mathbf{w} = [w_1, w_2, \dots, w_n]$.
- $b$: El sesgo (bias), que sigue siendo un único valor escalar.

##### La nueva ecuación del modelo

En lugar de $f_{w,b}(x) = w x + b$, ahora multiplicamos cada característica $x_j$ por su respectivo peso $w_j$ y sumamos el sesgo $b$:

$$f_{\mathbf{w},b}(\mathbf{x}) = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b$$

Expresión Vectorial (Producto Punto):


Aprovechando el álgebra lineal, podemos expresar toda esa suma como el producto punto ($\cdot$) entre el vector de pesos $\mathbf{w}$ y el vector de características $\mathbf{x}$:$$f_{\mathbf{w},b}(\mathbf{x}) = \mathbf{w} \cdot \mathbf{x} + b$$

Donde:

w = [w1, w2, ..., wn]

x = [x1, x2, ..., xn]

### Vectorización

La vectorización es una técnica de programación que permite realizar operaciones matemáticas sobre arreglos (arrays) enteros de una sola vez, sin necesidad de escribir bucles for explícitos en Python.

En lugar de procesar los datos elemento por elemento secuencialmente, NumPy aprovecha las instrucciones de procesamiento en paralelo a nivel de hardware (en la CPU usando instrucciones SIMD o en la GPU) para realizar los cálculos de forma simultánea.

1. Producto punto

El producto punto toma dos listas de números del mismo tamaño (dos vectores), multiplica los elementos que están en las mismas posiciones uno por uno, y suma todos los resultados.

El resultado final no es un vector, sino un solo número escalar.

$$\mathbf{w} \cdot \mathbf{x} = (w_1 \cdot x_1) + (w_2 \cdot x_2) + \dots + (w_n \cdot x_n)$$

Ejemplo práctico:

Imagina que quieres calcular el precio estimado de una casa basándote en $3$ características:

Vector de Características ($\mathbf{x}$):
- $x_1 = 120$ (metros cuadrados)
- $x_2 = 3$ (habitaciones)
- $x_3 = 10$ (años de antigüedad)

Vector de Pesos ($\mathbf{w}$):
- $w_1 = 2000$ (cada $\text{m}^2$ añade $\$2,000$)
- $w_2 = 10000$ (cada habitación añade $\$10,000$)
- $w_3 = -1500$ (cada año de antigüedad resta $\$1,500$)

El cálculo con el producto punto ($\mathbf{w} \cdot \mathbf{x}$):

$$\mathbf{w} \cdot \mathbf{x} = (2000 \times 120) + (10000 \times 3) + (-1500 \times 10)$$

$$\mathbf{w} \cdot \mathbf{x} = 240,000 + 30,000 - 15,000 = \mathbf{255,000}$$

Si sumamos el sesgo $b = 50,000$ (precio base del terreno), la predicción final es:

$$f_{\mathbf{w},b}(\mathbf{x}) = (\mathbf{w} \cdot \mathbf{x}) + b = 255,000 + 50,000 = \mathbf{305,000}$$

¿Cómo se hace en NumPy?

En Python, NumPy simplifica todo este proceso en una sola instrucción optimizada:

```python
import numpy as np

w = np.array([2000, 10000, -1500])
x = np.array([120, 3, 10])
b = 50000

# Cálculo del Producto Punto
prediccion = np.dot(w, x) + b

print(prediccion)  # Imprime: 305000
```

2. Ventaja en Rendimiento: Bucle for vs. Producto Punto

Comparativa Completa: Bucle for vs. Vectorización. 

Producto Punto Individual ($\mathbf{w} \cdot \mathbf{x}$)

Simulamos un caso con $n = 100,000$ características para ver el impacto en tiempo:

```python
import numpy as np
import time

# Fijar semilla para reproducibilidad
np.random.seed(42)

n = 100_000
w = np.random.rand(n)
x = np.random.rand(n)
b = 1.5

# -------------------------------------------------------------
# Opción A: Sin Vectorización (Bucle for en Python)
# -------------------------------------------------------------
tic = time.time()

f_wb_loop = 0.0
for j in range(n):
    f_wb_loop += w[j] * x[j]
f_wb_loop += b

toc = time.time()
time_loop = (toc - tic) * 1000  # Convertir a ms

print(f"Bucle for:    f_wb = {f_wb_loop:.4f} | Tiempo: {time_loop:.2f} ms")

# -------------------------------------------------------------
# Opción B: Con Vectorización (NumPy np.dot)
# -------------------------------------------------------------
tic = time.time()

f_wb_vec = np.dot(w, x) + b

toc = time.time()
time_vec = (toc - tic) * 1000  # Convertir a ms

print(f"Vectorizado:  f_wb = {f_wb_vec:.4f} | Tiempo: {time_vec:.2f} ms")
print(f"-> La vectorización fue ~{time_loop / time_vec:.1f}x más rápida.")
```

### Laboratorio: Regresión Lineal Múltiple Vectorizada

En este laboratorio implementaremos la Regresión Lineal Múltiple utilizando NumPy para vectorizar la función de costo y el cálculo del gradiente.

1. Importación de Librerías y Generación de Datos

Crearemos un dataset sintético con $m = 500$ ejemplos y $n = 3$ características ($x_1$: Tamaño $\text{m}^2$, $x_2$: Edad del inmueble, $x_3$: Distancia al centro en km) para predecir el precio en miles de dólares ($y$).

```python
import numpy as np
import matplotlib.pyplot as plt

# Fijar semilla para reproducibilidad
np.random.seed(42)

m = 500  # Número de ejemplos (filas)
n = 3    # Número de características (columnas)

# Generar datos sintéticos para X (m x n)
X = np.random.rand(m, n) * [100, 30, 15]  # Escalar para rangos realistas

# Pesos y sesgo "reales" que el modelo deberá aprender
w_true = np.array([2.5, -1.2, -0.8])
b_true = 50.0

# Generar Y agregando un poco de ruido gaussiano
y = np.dot(X, w_true) + b_true + np.random.randn(m) * 2.0

print(f"Dimensiones de X: {X.shape}")  # (500, 3)
print(f"Dimensiones de y: {y.shape}")  # (500,)
```

2. Función de Costo Vectorizada $J(\mathbf{w}, b)$

$$\mathbf{f} = \mathbf{X}\mathbf{w} + b$$

$$J(\mathbf{w},b) = \frac{1}{2m} \sum_{i=1}^{m} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})^2$$

```python
def compute_cost(X, y, w, b):
    """
    Calcula el costo MSE para Regresión Lineal Múltiple.
    
    Args:
      X (ndarray (m,n)): Datos de entrada, m ejemplos con n características
      y (ndarray (m,)):  Valores objetivo reales
      w (ndarray (n,)):  Vector de pesos del modelo
      b (scalar):        Término de sesgo
      
    Returns:
      cost (float): Costo total J(w,b)
    """
    m = X.shape[0]
    
    # 1. Calcular predicciones para todos los ejemplos (vector de tamaño m)
    f_wb = np.dot(X, w) + b
    
    # 2. Calcular los errores al cuadrado y promediar
    cost = np.sum((f_wb - y) ** 2) / (2 * m)
    
    return cost
```

3. Función de Gradiente Vectorizada

$$\frac{\partial J}{\partial w_j} = \frac{1}{m} \sum_{i=1}^{m} \left( f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)} \right) \cdot x_j^{(i)}$$

$$\frac{\partial J}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} (f_{\mathbf{w},b}(\mathbf{x}^{(i)}) - y^{(i)})$$

```python
def compute_gradient(X, y, w, b):
    m, n = X.shape
    
    # Inicializamos las derivadas en cero
    dj_dw = np.zeros(n)  # Un gradiente para cada feature (tamaño n)
    dj_db = 0.0          # Un solo gradiente para el sesgo
    
    # Recorremos cada uno de los m ejemplos
    for i in range(m):
        # 1. Error de predicción del ejemplo i
        err = (np.dot(X[i], w) + b) - y[i]
        
        # 2. Acumulamos el gradiente para el sesgo b
        dj_db += err
        
        # 3. Acumulamos el gradiente para cada peso w_j
        for j in range(n):
            dj_dw[j] += err * X[i, j]
            
    # Sacamos el promedio dividiendo entre m
    dj_dw = dj_dw / m
    dj_db = dj_db / m
    
    return dj_dw, dj_db
```

4. Algoritmo de Descenso del Gradiente

```python
def gradient_descent(X, y, w_in, b_in, alpha, num_iters):
    """
    Ejecuta el descenso del gradiente para aprender w y b.
    """
    w = w_in.copy()
    b = b_in
    J_history = []
    
    for i in range(num_iters):
        # Calcular gradientes
        dj_dw, dj_db = compute_gradient(X, y, w, b)
        
        # Actualización simultánea de parámetros
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        
        # Guardar costo en cada iteración
        if i < 100000:  # Prevenir sobrecarga de memoria
            J_history.append(compute_cost(X, y, w, b))
            
        # Imprimir progreso cada 10% de iteraciones
        if i % (num_iters // 10) == 0:
            print(f"Iteración {i:5d}: Costo = {J_history[-1]:.4f}")
            
    return w, b, J_history
```

5. Entrenamiento del Modelo

```python
# Inicializar parámetros en cero
w_init = np.zeros(n)
b_init = 0.0

# Hiperparámetros (Alpha ajustado debido a las escalas de las variables)
iterations = 1000
alpha = 0.0001

# Ejecutar el entrenamiento
w_final, b_final, J_hist = gradient_descent(X, y, w_init, b_init, alpha, iterations)

print("\n--- Resultados del Entrenamiento ---")
print(f"w verdadero: {w_true} | w aprendido: {np.round(w_final, 2)}")
print(f"b verdadero: {b_true} | b aprendido: {b_final:.2f}")
```

### Conclusion

Observación Técnica: Al ejecutar el Descenso del Gradiente sobre nuestro modelo de Regresión Lineal Múltiple, se observa que el costo $J(\mathbf{w}, b)$ se estanca o requiere un $\alpha$ extremadamente pequeño ($0.0001$) para no divergir, resultando en predicciones imprecisas y un aprendizaje muy lento.

¿Por qué sucede esto?Debido a las diferentes escalas de las características ($x_1$ está en cientos de $\text{m}^2$, mientras que $x_2$ está en decenas de años). Esto provoca que la superficie del costo sea extremadamente alargada (un "tazón deformado"), haciendo que el gradiente oscile de forma ineficiente.
