## Regresión lineal múltiple

En la vida real, una sola variable rara vez es suficiente para predecir un resultado con precisión. Por ejemplo, al predecir el precio de una casa no solo observamos su tamaño ($x$), sino también el número de habitaciones, los años de antigüedad y el número de pisos.

Cuando el modelo utiliza dos o más características de entrada (features), pasamos de la Regresión Lineal Simple a la Regresión Lineal Múltiple.

#### Nueva notación matemática

Para manejar múltiples variables sin confundirnos, adoptamos la siguiente convención matemática estándar:
- $n$: Número total de características (features). Por ejemplo, $n = 4$ si usamos (Tamaño, Habitaciones, Pisos, Antigüedad).
- $m$: Número total de ejemplos de entrenamiento (filas del dataset).
- $\mathbf{x}^{(i)}$: Un vector fila que contiene todas las características del ejemplo $i$-ésimo. Por ejemplo, $\mathbf{x}^{(2)} = [120, 3, 2, 10]$.
- $x_j^{(i)}$: El valor de la característica $j$ en el ejemplo $i$. Por ejemplo, $x_3^{(2)} = 2$ (el número de pisos de la casa 2).
- $\mathbf{w}$: Vector de pesos (weights), donde cada $w_j$ es un número que acompaña a la característica $x_j$. Se representa como $\mathbf{w} = [w_1, w_2, \dots, w_n]$.
- $b$: El sesgo (bias), que sigue siendo un único valor escalar.

##### La nueva ecuación del modelo

En lugar de $f_{w,b}(x) = w x + b$, ahora multiplicamos cada característica $x_j$ por su respectivo peso $w_j$ y sumamos el sesgo $b$:

$$f_{\mathbf{w},b}(\mathbf{x}) = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b$$

Expresión Vectorial (Producto Punto):


Aprovechando el álgebra lineal, podemos expresar toda esa suma como el producto punto ($\cdot$) entre el vector de pesos $\mathbf{w}$ y el vector de características $\mathbf{x}$:$$f_{\mathbf{w},b}(\mathbf{x}) = \mathbf{w} \cdot \mathbf{x} + b$$

Donde:

w = [w1, w2, ..., wn]

x = [x1, x2, ..., xn]
